```python
import pandas as pd
import numpy as np
from faker import Faker
import random

# Set seed for reproducibility
random.seed(42)
np.random.seed(42)

fake = Faker()
```

```python
# Number of transactions
NUM_TRANSACTIONS = 10000
```

```python
countries = [
    "UAE",
    "Saudi Arabia",
    "Kuwait",
    "Qatar"
]

merchants = [
    "Noon",
    "Amazon UAE",
    "Jarir",
    "Namshi",
    "Extra",
    "Carrefour",
    "IKEA",
    "Centrepoint",
    "Sharaf DG",
    "Virgin Megastore"
]

categories = [
    "Electronics",
    "Fashion",
    "Home",
    "Beauty",
    "Travel"
]

payment_status = [
    "Paid",
    "Late",
    "Default"
]

installments = [2,3,4,6]
```

```python
transactions = []

for i in range(NUM_TRANSACTIONS):

    transactions.append({

        "Transaction_ID": i + 1,

        "Customer_ID": random.randint(100000,999999),

        "Customer_Name": fake.name(),

        "Email": fake.email(),

        "Phone": fake.phone_number(),

        "Country": random.choice(countries),

        "Merchant": random.choice(merchants),

        "Category": random.choice(categories),

        "Order_Value": round(random.uniform(50,5000),2),

        "Installments": random.choice(installments),

        "Payment_Status": random.choices(
            payment_status,
            weights=[80,15,5]
        )[0],

        "Days_Late": random.randint(0,45),

        "Credit_Score": random.randint(550,850),

        "Monthly_Income": random.randint(3000,30000),

        "Age": random.randint(18,60),

        "Order_Date": fake.date_between(
            start_date="-2y",
            end_date="today"
        )

    })
```

```python
df = pd.DataFrame(transactions)
```

```python
df.loc[df.sample(150).index, "Monthly_Income"] = np.nan

df.loc[df.sample(120).index, "Credit_Score"] = np.nan
```

```python
duplicates = df.sample(75)

df = pd.concat([df, duplicates], ignore_index=True)
```

```python
df.loc[df.sample(40).index, "Order_Value"] = -100
```

```python
df.loc[df.sample(20).index, "Merchant"] = "Nooon"

df.loc[df.sample(20).index, "Merchant"] = "Amazon"

df.loc[df.sample(20).index, "Merchant"] = "Jarirr"
```

```python
print("Shape:", df.shape)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows")
print(df.duplicated().sum())
```

```python
df.to_csv("raw_bnpl_data.csv", index=False)

print("Raw dataset saved successfully!")
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("raw_bnpl_data.csv")

print(df.shape)
df.head()
```

```python
print("Data Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("New Shape:", df.shape)
```

```python
df["Monthly_Income"] = df["Monthly_Income"].fillna(df["Monthly_Income"].median())

df["Credit_Score"] = df["Credit_Score"].fillna(df["Credit_Score"].median())
```

```python
df.isnull().sum()
```

```python
df[df["Order_Value"] < 0]
```

```python
median_order = df[df["Order_Value"] > 0]["Order_Value"].median()

df.loc[df["Order_Value"] < 0, "Order_Value"] = median_order
```

```python
merchant_mapping = {
    "Amazon": "Amazon UAE",
    "Nooon": "Noon",
    "Jarirr": "Jarir"
}

df["Merchant"] = df["Merchant"].replace(merchant_mapping)
```

```python
df["Merchant"].value_counts()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df["Order_Month"] = df["Order_Date"].dt.month_name()
```

```python
df["Order_Year"] = df["Order_Date"].dt.year
```

```python
df["Quarter"] = df["Order_Date"].dt.quarter
```

```python
conditions = [
    df["Credit_Score"] >= 720,
    (df["Credit_Score"] >= 650) & (df["Credit_Score"] < 720),
    df["Credit_Score"] < 650
]

choices = [
    "Low Risk",
    "Medium Risk",
    "High Risk"
]

df["Risk_Level"] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```

```python
conditions = [
    df["Age"] <= 25,
    (df["Age"] > 25) & (df["Age"] <= 40),
    df["Age"] > 40
]

choices = [
    "18-25",
    "26-40",
    "40+"
]

df["Age_Group"] = np.select(conditions, choices)
```

```python
conditions = [
    df["Age"] <= 25,
    (df["Age"] > 25) & (df["Age"] <= 40),
    df["Age"] > 40
]

choices = [
    "18-25",
    "26-40",
    "40+"
]

df["Age_Group"] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```

```python
conditions = [
    df["Monthly_Income"] < 8000,
    (df["Monthly_Income"] >= 8000) & (df["Monthly_Income"] < 15000),
    df["Monthly_Income"] >= 15000
]

choices = [
    "Low Income",
    "Middle Income",
    "High Income"
]

df["Income_Band"] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```

```python
conditions = [
    df["Order_Value"] >= 3000,
    (df["Order_Value"] >= 1000) & (df["Order_Value"] < 3000),
    df["Order_Value"] < 1000
]

choices = [
    "Premium",
    "Regular",
    "Budget"
]

df["Customer_Segment"] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```

```python
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows")
print(df.duplicated().sum())

df.head()
```

```python
df.to_csv("clean_bnpl_data.csv", index=False)

print("Clean dataset exported successfully!")
```
